#### ResNet-50

```python
# ResNet50 for content and composition
model = models.resnet50(pretrained=True)

# Remove final classification layer (fc)
model = torch.nn.Sequential(*list(model.children())[:-1])
model.to(device)
model.eval()  

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std =[0.229, 0.224, 0.225]
    )
])
```

```python
def process_dataset_batch(image_root, batch_size=32):
    image_tensors = []
    metadata = []  

    all_embeddings = []
    all_metadata = []

    for root, dirs, files in os.walk(image_root):
        for filename in files:
            if not filename.endswith(('.jpg', '.png')):
                continue

            path = os.path.join(root, filename)

            try:
                genre = os.path.basename(root)

                image = Image.open(path).convert('RGB')
                image = transform(image)

                image_tensors.append(image)
                metadata.append((filename, genre))

            except Exception as e:
                print(f"Skipping {filename}: {e}")
                continue

            # Process batch
            if len(image_tensors) == batch_size:
                emb, meta = process_batch(image_tensors, metadata)
                all_embeddings.append(emb)
                all_metadata.extend(meta)

                image_tensors, metadata = [], []

    # Process remaining
    if image_tensors:
        emb, meta = process_batch(image_tensors, metadata)
        all_embeddings.append(emb)
        all_metadata.extend(meta)

    # Combine all batches
    all_embeddings = np.vstack(all_embeddings)

    return all_embeddings, all_metadata

def process_batch(images, metadata):
    batch = torch.stack(images).to(device)

    with torch.no_grad():
        embeddings = model(batch)

    embeddings = embeddings.squeeze(-1).squeeze(-1)  # (B, 2048)
    embeddings = embeddings.cpu().numpy()

    # Print per painting
    for i, ((filename, genre), emb) in enumerate(zip(metadata, embeddings)):
        print(f"[{i}] Processed: {filename} | Genre: {genre} | Embedding shape: {emb.shape}")

    return embeddings, metadata

def save_embeddings(output_path, embeddings, metadata):
    filenames = [m[0] for m in metadata]
    genres = [m[1] for m in metadata]

    np.savez(
        output_path,
        embeddings=embeddings,
        filenames=filenames,
        genres=genres
    )

# def insert_batch_embeddings(conn, painting_ids, embeddings):
#     with conn.cursor() as cur:
#         data = [
#             (pid, emb.tolist())
#             for pid, emb in zip(painting_ids, embeddings)
#         ]

#         cur.executemany("""
#             INSERT INTO painting_cnn_embeddings (painting_id, resnet50_embedding)
#             VALUES (%s, %s)
#         """, data)

#     conn.commit()

# def get_painting_id_from_filename(conn, filename):
#     with conn.cursor() as cur:
#         cur.execute("""
#             SELECT painting_id 
#             FROM paintings 
#             WHERE image_filename = %s
#         """, (filename,))
#         result = cur.fetchone()

#         if result is None:
#             raise ValueError(f"No DB entry for {filename}")

#         return result[0]
    
# def load_filename_to_id_map(conn):
#     with conn.cursor() as cur:
#         cur.execute("SELECT painting_id, image_filename FROM paintings")
#         rows = cur.fetchall()

#     return {filename: pid for pid, filename in rows}

# filename_to_id = load_filename_to_id_map(conn)
# painting_id = filename_to_id[filename]
```

#### VGG19

```python
# VGG19 for style and texture 
image = r'C:\Users\danie\Desktop\_\Daniela Curmi\University\Final Year Project\Final Year Project Code Implementation\Website\paintings\Abstract_Expressionism\aaron-siskind_acolman-1-1955.jpg'

# Load base model without classifier
base_model = VGG19(weights='imagenet', include_top=False)

# Choose multiple layers (better for art representation)
layer_names = [
    'block3_conv4',  # texture
    'block4_conv4',  # style
    'block5_conv4'   # semantic
]

outputs = [base_model.get_layer(name).output for name in layer_names]

# Apply pooling to each
pooled_outputs = [GlobalAveragePooling2D()(out) for out in outputs]

model_vgg = Model(inputs=base_model.input, outputs=pooled_outputs)

def extract_vgg19_features(img):
    img = cv2.resize(img, (224, 224))
    img = preprocess_input(img.astype(np.float32))
    img = np.expand_dims(img, axis=0)

    features = model_vgg.predict(img)

    # Concatenate multi-level features
    feature_vector = np.concatenate(features, axis=1)

    return feature_vector.flatten()

def visualize_feature_maps(img, layer_name='block1_conv1', max_maps=64):
    model = Model(inputs=base_model.input,
                  outputs=base_model.get_layer(layer_name).output)

    img_resized = cv2.resize(img, (224, 224))
    img_processed = preprocess_input(img_resized.astype(np.float32))
    img_processed = np.expand_dims(img_processed, axis=0)

    feature_maps = model.predict(img_processed)[0]

    n_maps = min(max_maps, feature_maps.shape[-1])
    size = int(np.sqrt(n_maps))

    plt.figure(figsize=(10,10))

    for i in range(n_maps):
        plt.subplot(size, size, i+1)
        plt.imshow(feature_maps[:, :, i], cmap='viridis')
        plt.axis('off')

    plt.suptitle(f"Feature maps: {layer_name}")
    plt.show()

def visualize_filters(layer_name='block1_conv1', n_filters=6):
    layer = base_model.get_layer(layer_name)
    filters, biases = layer.get_weights()

    # Normalize to 0–1
    f_min, f_max = filters.min(), filters.max()
    filters = (filters - f_min) / (f_max - f_min)

    plt.figure(figsize=(10, 6))

    for i in range(n_filters):
        f = filters[:, :, :, i]

        for j in range(3):  # RGB channels
            plt.subplot(n_filters, 3, i*3 + j + 1)
            plt.imshow(f[:, :, j], cmap='gray')
            plt.axis('off')

    plt.suptitle(f"Filters: {layer_name}")
    plt.show()

extract_vgg19_features(image)
```

```python
visualize_feature_maps(image, 'block1_conv1')  # edges
visualize_feature_maps(image, 'block3_conv4')  # textures
visualize_feature_maps(image, 'block5_conv4')  # abstract patterns
```

### Gram Matrices for Style Features

The style represents the texture and pattern of a painting e.g. the type of brush stroke used or patterns in a painting


#### Gram Matrix

```python
# Preprocessing pipeline - normalise as VGG expects
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

def load_image(image_path):
    image = Image.open(image_path).convert("RGB")
    image = transform(image).unsqueeze(0)  # Add batch dimension
    return image

painting_tensor = load_image(img_path)

def gram_matrix(features):
    # features: [batch, channels, height, width]
    b, c, h, w = features.size()
    features = features.view(c, h * w)  # flatten spatial dimensions
    gram = features @ features.t()      # standard matrix multiplication
    gram /= (c * h * w)                 # normalize
    return gram

vgg = models.vgg19(pretrained=True).features.eval()

# Specific layers that capture style 
style_layers = ['0', '5', '10', '19', '28']

def extract_gram_features(image, model, style_layers):
    grams = []
    x = image
    for name, layer in model._modules.items():
        x = layer(x)
        if name in style_layers:
            grams.append(gram_matrix(x))
    return grams

with torch.no_grad():
    gram_features = extract_gram_features(painting_tensor, vgg, style_layers)

print(gram_features)
```
